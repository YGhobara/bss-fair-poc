```python
profile = pd.DataFrame({
    "column": df.columns,
    "non_null": [df[c].notna().sum() for c in df.columns],
    "missing_pct": [df[c].isna().mean() * 100 for c in df.columns],
    "unique": [df[c].nunique(dropna=True) for c in df.columns],
})

profile.sort_values("missing_pct")
```

```python
profile["missing_pct"] = profile["missing_pct"].round(2)

profile.sort_values(
    "missing_pct"
).reset_index(drop=True)
```

```python
profile.sort_values(
    ["unique", "missing_pct"]
).reset_index(drop=True)
```

```python
print(profile.sort_values("missing_pct").to_string(index=False))
```

```python
for col in lex_columns:
    print(f"\n--- {col} ---")
    print(df[col].value_counts(dropna=False).head(15))
```

```python
r = df["lex_recherche"].notna()
rec = df["lex_reconnaissance"].notna()

print("Recherche only:", (r & ~rec).sum())
print("Reconnaissance only:", (~r & rec).sum())
print("Both:", (r & rec).sum())
print("Neither:", (~r & ~rec).sum())
```
